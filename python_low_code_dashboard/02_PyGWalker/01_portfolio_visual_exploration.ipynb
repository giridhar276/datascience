# Portfolio Visual Exploration

Use PyGWalker as a drag-and-drop visual analysis environment. Suggested explorations:
- revenue by country and industry;
- profitability versus market share;
- cloud platform versus technology adoption;
- high-risk company drill-down.

```python
import pandas as pd
import pygwalker as pyg
from pathlib import Path

data_file = Path("data/software_companies_dataset_v2.csv")
df = pd.read_csv(data_file)

# Prepare missing values for reliable widgets and Plotly charts.
categorical_columns = [
    "Company_Name", "Industry", "Headquarters_City", "Country",
    "Ownership_Type", "Customer_Segment", "Primary_Cloud", "Risk_Rating"
]
for column in categorical_columns:
    if column in df.columns:
        df[column] = df[column].fillna("Unknown").astype(str).str.strip()

numeric_columns = [
    "Employees", "Annual_Revenue", "Profit_Margin", "Market_Share",
    "R&D_Spending", "Average_Salary", "Training_Hours_Per_Employee",
    "Employee_Satisfaction", "Adoption_Rate_AI", "Adoption_Rate_Cloud",
    "Adoption_Rate_Blockchain"
]
for column in numeric_columns:
    if column in df.columns:
        df[column] = pd.to_numeric(df[column], errors="coerce")
        df[column] = df[column].fillna(df[column].median())

df["Annual_Revenue"] = df["Annual_Revenue"].clip(lower=1)
df["Employees"] = df["Employees"].clip(lower=1)

df["Incorporation_Date"] = pd.to_datetime(df["Incorporation_Date"], errors="coerce")
df["Last_Funding_Date"] = pd.to_datetime(df["Last_Funding_Date"], errors="coerce")
df.head()
```

```python
walker = pyg.walk(
    df,
    env="JupyterAnywidget",
    appearance="light"
)
```
