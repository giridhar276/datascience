# Engineered Metrics Exploration

Create calculated business indicators before launching the visual explorer. This demonstrates that PyGWalker can explore both raw and engineered fields.

```python
import pandas as pd
import numpy as np
import pygwalker as pyg

df = pd.read_csv("data/software_companies_dataset_v2.csv")

# Prepare missing values for reliable widgets and Plotly charts.
categorical_columns = [
    "Company_Name", "Industry", "Headquarters_City", "Country",
    "Ownership_Type", "Customer_Segment", "Primary_Cloud", "Risk_Rating"
]
for column in categorical_columns:
    if column in df.columns:
        df[column] = df[column].fillna("Unknown").astype(str).str.strip()

numeric_columns = [
    "Employees", "Annual_Revenue", "Profit_Margin", "Market_Share",
    "R&D_Spending", "Average_Salary", "Training_Hours_Per_Employee",
    "Employee_Satisfaction", "Adoption_Rate_AI", "Adoption_Rate_Cloud",
    "Adoption_Rate_Blockchain"
]
for column in numeric_columns:
    if column in df.columns:
        df[column] = pd.to_numeric(df[column], errors="coerce")
        df[column] = df[column].fillna(df[column].median())

df["Annual_Revenue"] = df["Annual_Revenue"].clip(lower=1)
df["Employees"] = df["Employees"].clip(lower=1)

df["Revenue_Per_Employee"] = df["Annual_Revenue"] / df["Employees"].replace(0, np.nan)
df["R&D_Intensity"] = 100 * df["R&D_Spending"] / df["Annual_Revenue"].replace(0, np.nan)
df["Technology_Index"] = df[
    ["Adoption_Rate_AI", "Adoption_Rate_Cloud", "Adoption_Rate_Blockchain"]
].mean(axis=1)
df["Estimated_Profit"] = df["Annual_Revenue"] * df["Profit_Margin"] / 100
df["Company_Age"] = pd.Timestamp.today().year - df["Year_Founded"]
df[["Revenue_Per_Employee", "R&D_Intensity", "Technology_Index", "Estimated_Profit"]].describe()
```

```python

pyg.walk(df)
```
