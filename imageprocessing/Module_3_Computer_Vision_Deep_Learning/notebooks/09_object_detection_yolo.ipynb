# Use Case 9: Object Detection using a Pretrained YOLO Model

Object detection predicts object classes, confidence scores and bounding boxes.

Each code cell is preceded by Markdown that explains what the step does, why it is needed, and which deep-learning concept is being demonstrated.


## Step 1: Install and import Ultralytics

The first execution may install the library and download pretrained weights.

```python
# Uncomment if required:
# !pip install ultralytics

from pathlib import Path
from ultralytics import YOLO
```

## Step 2: Locate an included real image

A real photograph is already available in the ZIP.

```python
IMAGE_PATH = Path(
    "../datasets/representative_real_images/astronaut.png"
)

print(IMAGE_PATH.resolve())
```

## Step 3: Load a pretrained detector

`yolo11n.pt` is a small and fast pretrained detector.

```python
model = YOLO("yolo11n.pt")

print("Model loaded.")
```

## Step 4: Run detection

`conf=0.25` filters weak predictions. `save=True` writes an annotated output image.

```python
results = model.predict(
    source=str(IMAGE_PATH),
    conf=0.25,
    save=True,
)

print("Results:", len(results))
```

## Step 5: Inspect predictions

The boxes object contains coordinates, class identifiers and confidence scores.

```python
for result in results:
    print(result.boxes)
    print("Annotated image directory:", result.save_dir)
```
