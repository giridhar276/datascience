# Tree-Based Classification Models

This notebook demonstrates three supervised classification algorithms:

1. Decision Tree Classification
2. Random Forest Classification
3. Gradient Boosting Classification

Two datasets are used:

- **Car Accidents IoT dataset** for Decision Tree and Random Forest
- **E-commerce Products Menu dataset** for Decision Tree and Gradient Boosting

The notebook includes:

- Dataset loading
- Data inspection
- Feature and target selection
- Model training
- Test-data prediction
- Confusion matrix
- Accuracy and F1-score
- Model comparison

> Make sure the datasets are available in the paths used in this notebook.


## 1. Import Required Libraries

The following libraries are used:

- `pandas` for reading and manipulating datasets
- `numpy` for numerical operations
- `matplotlib` for visualizations
- `scikit-learn` for machine learning models and evaluation metrics
- `time` for measuring model training duration


````python
import time
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.tree import DecisionTreeClassifier, plot_tree
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier
from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay,
    f1_score
)

pd.set_option("display.max_columns", None)
````

# Part 1: Car Accident Classification

The first dataset is used to predict whether an accident is fatal.

- Training dataset: `Car Accidents IOT/train.csv`
- Testing dataset: `Car Accidents IOT/test.csv`
- Target column: `Fatal`


## 2. Define Dataset Paths

Using `Path` makes file paths easier to read and manage.

Update the paths below when the datasets are stored in another folder.


````python
car_train_path = Path("./Datasets/Car Accidents IOT/train.csv")
car_test_path = Path("./Datasets/Car Accidents IOT/test.csv")

print("Training file:", car_train_path)
print("Testing file :", car_test_path)
````

## 3. Load the Car Accident Datasets

The training data is used to build the models.

The testing data is used to evaluate model performance on unseen records.


````python
car_train = pd.read_csv(car_train_path)
car_test = pd.read_csv(car_test_path)

print("Training data loaded successfully.")
print("Testing data loaded successfully.")
````

## 4. Preview the Datasets

Display the first few rows of both datasets.


````python
print("Car training data:")
display(car_train.head())

print("Car testing data:")
display(car_test.head())
````

## 5. Inspect Dataset Structure

The `info()` method shows:

- Number of records
- Column names
- Data types
- Missing values
- Memory usage


````python
print("Training dataset information:")
car_train.info()

print("\nTesting dataset information:")
car_test.info()
````

## 6. Check Dataset Shapes and Missing Values


````python
print("Training dataset shape:", car_train.shape)
print("Testing dataset shape :", car_test.shape)

print("\nMissing values in training data:")
display(car_train.isnull().sum())

print("\nMissing values in testing data:")
display(car_test.isnull().sum())
````

## 7. Define Features and Target

The original code selects columns from position 1 through 21.

```python
var = list(car_train.columns[1:22])
```

These selected columns become the input features.

The target column is `Fatal`.


````python
car_features = list(car_train.columns[1:22])

X_car_train = car_train[car_features]
y_car_train = car_train["Fatal"]

X_car_test = car_test[car_features]
y_car_test = car_test["Fatal"]

print("Selected feature columns:")
print(car_features)

print("\nTraining feature shape:", X_car_train.shape)
print("Testing feature shape :", X_car_test.shape)
````

## 8. Inspect the Target Distribution

The distribution shows how many fatal and non-fatal records are available.


````python
print("Training target distribution:")
display(y_car_train.value_counts())

print("Training target percentage:")
display(y_car_train.value_counts(normalize=True).mul(100).round(2))
````

# Model 1: Decision Tree Classifier

A Decision Tree creates decision rules by repeatedly splitting the data.

The default model is first trained without restricting its depth or number of leaves.


````python
decision_tree_model = DecisionTreeClassifier(random_state=42)

decision_tree_model.fit(X_car_train, y_car_train)

print("Decision Tree model trained successfully.")
````

## 9. Predict Using the Decision Tree


````python
tree_predictions = decision_tree_model.predict(X_car_test)

print("First 10 Decision Tree predictions:")
print(tree_predictions[:10])
````

## 10. Decision Tree Confusion Matrix

The confusion matrix summarizes correct and incorrect predictions.

For a binary classification problem, it normally contains:

- True Negatives
- False Positives
- False Negatives
- True Positives


````python
tree_cm = confusion_matrix(y_car_test, tree_predictions)

print("Decision Tree Confusion Matrix:")
print(tree_cm)
````

## 11. Calculate Decision Tree Accuracy

The original code manually calculates accuracy from the confusion matrix.

This notebook shows both:

- Manual calculation
- `accuracy_score()` calculation


````python
tree_total = tree_cm.sum()

if tree_cm.shape == (2, 2):
    tree_accuracy_manual = (tree_cm[0, 0] + tree_cm[1, 1]) / tree_total
else:
    tree_accuracy_manual = np.trace(tree_cm) / tree_total

tree_accuracy = accuracy_score(y_car_test, tree_predictions)

print(f"Decision Tree Accuracy - Manual Calculation: {tree_accuracy_manual:.4f}")
print(f"Decision Tree Accuracy - accuracy_score():    {tree_accuracy:.4f}")
````

## 12. Decision Tree Classification Report


````python
print("Decision Tree Classification Report:")
print(classification_report(y_car_test, tree_predictions))
````

## 13. Display the Decision Tree Confusion Matrix


````python
ConfusionMatrixDisplay.from_predictions(
    y_car_test,
    tree_predictions
)

plt.title("Decision Tree Confusion Matrix")
plt.show()
````

# Model 2: Random Forest Classifier

A Random Forest combines predictions from multiple Decision Trees.

Original parameters:

- `n_estimators=10`: Build 10 Decision Trees.
- `max_features=5`: Consider up to five features while searching for the best split.

A fixed `random_state` is added for reproducible results.


````python
random_forest_model = RandomForestClassifier(
    n_estimators=10,
    max_features=5,
    random_state=42,
    n_jobs=-1
)

random_forest_model.fit(X_car_train, y_car_train)

print("Random Forest model trained successfully.")
````

## 14. Predict Using the Random Forest


````python
forest_predictions = random_forest_model.predict(X_car_test)

print("First 10 Random Forest predictions:")
print(forest_predictions[:10])
````

## 15. Random Forest Confusion Matrix


````python
forest_cm = confusion_matrix(y_car_test, forest_predictions)

print("Random Forest Confusion Matrix:")
print(forest_cm)
````

## 16. Calculate Random Forest Accuracy


````python
forest_total = forest_cm.sum()

if forest_cm.shape == (2, 2):
    forest_accuracy_manual = (
        forest_cm[0, 0] + forest_cm[1, 1]
    ) / forest_total
else:
    forest_accuracy_manual = np.trace(forest_cm) / forest_total

forest_accuracy = accuracy_score(y_car_test, forest_predictions)

print(f"Random Forest Accuracy - Manual Calculation: {forest_accuracy_manual:.4f}")
print(f"Random Forest Accuracy - accuracy_score():    {forest_accuracy:.4f}")
````

## 17. Random Forest Classification Report


````python
print("Random Forest Classification Report:")
print(classification_report(y_car_test, forest_predictions))
````

## 18. Compare Decision Tree and Random Forest


````python
car_model_comparison = pd.DataFrame({
    "Model": ["Decision Tree", "Random Forest"],
    "Accuracy": [tree_accuracy, forest_accuracy],
    "F1 Score - Weighted": [
        f1_score(y_car_test, tree_predictions, average="weighted"),
        f1_score(y_car_test, forest_predictions, average="weighted")
    ]
})

car_model_comparison
````

## 19. Random Forest Feature Importance

Feature importance indicates which variables contributed most to the Random Forest predictions.


````python
feature_importance = pd.DataFrame({
    "Feature": car_features,
    "Importance": random_forest_model.feature_importances_
}).sort_values("Importance", ascending=False)

feature_importance
````

# Part 2: Product Category Classification

The second dataset is used to predict an e-commerce product category.

- Training dataset: `Ecom_Products_Menu/train.csv`
- Testing dataset: `Ecom_Products_Menu/test.csv`
- Target column: `Category`

Two models are built:

1. Decision Tree
2. Gradient Boosting


## 20. Define the Product Dataset Paths


````python
menu_train_path = Path("./Datasets/Ecom_Products_Menu/train.csv")
menu_test_path = Path("./Datasets/Ecom_Products_Menu/test.csv")

print("Training file:", menu_train_path)
print("Testing file :", menu_test_path)
````

## 21. Load the Product Datasets


````python
menu_train = pd.read_csv(menu_train_path)
menu_test = pd.read_csv(menu_test_path)

print("Product training data loaded successfully.")
print("Product testing data loaded successfully.")
````

## 22. Preview and Inspect the Product Datasets


````python
print("Menu training data:")
display(menu_train.head())

print("Menu testing data:")
display(menu_test.head())

print("Training shape:", menu_train.shape)
print("Testing shape :", menu_test.shape)
````

## 23. Define Product Features and Target

The original code selects columns from position 1 through 100.

```python
lab = list(menu_train.columns[1:101])
```

These columns are used as input features.

The target column is `Category`.


````python
menu_features = list(menu_train.columns[1:101])

X_menu_train = menu_train[menu_features]
y_menu_train = menu_train["Category"]

X_menu_test = menu_test[menu_features]
y_menu_test = menu_test["Category"]

print("Number of selected features:", len(menu_features))
print("Training feature shape:", X_menu_train.shape)
print("Testing feature shape :", X_menu_test.shape)
````

## 24. Inspect Product Category Distribution


````python
menu_train["Category"].value_counts()
````

# Model 3: Decision Tree for Product Classification


````python
menu_tree_model = DecisionTreeClassifier(random_state=42)

menu_tree_model.fit(X_menu_train, y_menu_train)

print("Product Decision Tree model trained successfully.")
````

## 25. Predict Product Categories


````python
menu_tree_predictions = menu_tree_model.predict(X_menu_test)

print("First 10 predictions:")
print(menu_tree_predictions[:10])
````

## 26. Product Decision Tree Confusion Matrix


````python
menu_tree_cm = confusion_matrix(y_menu_test, menu_tree_predictions)

print("Product Decision Tree Confusion Matrix:")
print(menu_tree_cm)
````

## 27. Product Decision Tree F1-Score

`average='micro'` calculates the metric globally by counting all true positives, false negatives, and false positives.

For single-label multiclass classification, micro F1-score is generally equal to accuracy.


````python
menu_tree_f1 = f1_score(
    y_menu_test,
    menu_tree_predictions,
    average="micro"
)

menu_tree_accuracy = accuracy_score(
    y_menu_test,
    menu_tree_predictions
)

print(f"Product Decision Tree Micro F1-Score: {menu_tree_f1:.4f}")
print(f"Product Decision Tree Accuracy:       {menu_tree_accuracy:.4f}")
````

# Model 4: Gradient Boosting Classifier

Gradient Boosting builds models sequentially.

Each new model focuses more on correcting errors made by the previous models.

### Parameters

- `loss='log_loss'`: Uses logarithmic loss for classification.
- `learning_rate=0.1`: Controls the contribution of each new tree.
- `n_estimators=100`: Builds 100 boosting stages.
- `min_samples_leaf=5`: Requires at least five records in a leaf.
- `max_depth=3`: Limits each individual tree's depth.
- `verbose=1`: Displays model training progress.


````python
gradient_boosting_model = GradientBoostingClassifier(
    loss="log_loss",
    learning_rate=0.1,
    n_estimators=100,
    min_samples_leaf=5,
    max_depth=3,
    verbose=1,
    random_state=42
)
````

## 28. Train the Gradient Boosting Model and Measure Time

`time.perf_counter()` is used to measure the training duration accurately.


````python
start_time = time.perf_counter()

gradient_boosting_model.fit(X_menu_train, y_menu_train)

end_time = time.perf_counter()
gradient_boosting_training_time = end_time - start_time

print(f"Gradient Boosting Training Time: {gradient_boosting_training_time:.4f} seconds")
````

## 29. Predict Using Gradient Boosting


````python
boost_predictions = gradient_boosting_model.predict(X_menu_test)

print("First 10 Gradient Boosting predictions:")
print(boost_predictions[:10])
````

## 30. Gradient Boosting Confusion Matrix


````python
boost_cm = confusion_matrix(y_menu_test, boost_predictions)

print("Gradient Boosting Confusion Matrix:")
print(boost_cm)
````

## 31. Gradient Boosting Accuracy and F1-Score


````python
boost_accuracy = accuracy_score(y_menu_test, boost_predictions)

boost_f1_micro = f1_score(
    y_menu_test,
    boost_predictions,
    average="micro"
)

boost_f1_weighted = f1_score(
    y_menu_test,
    boost_predictions,
    average="weighted"
)

print(f"Gradient Boosting Accuracy:          {boost_accuracy:.4f}")
print(f"Gradient Boosting Micro F1-Score:    {boost_f1_micro:.4f}")
print(f"Gradient Boosting Weighted F1-Score: {boost_f1_weighted:.4f}")
````

## 32. Gradient Boosting Classification Report


````python
print("Gradient Boosting Classification Report:")
print(classification_report(y_menu_test, boost_predictions))
````

## 33. Compare Product Classification Models


````python
menu_model_comparison = pd.DataFrame({
    "Model": ["Decision Tree", "Gradient Boosting"],
    "Accuracy": [
        menu_tree_accuracy,
        boost_accuracy
    ],
    "Micro F1-Score": [
        menu_tree_f1,
        boost_f1_micro
    ],
    "Weighted F1-Score": [
        f1_score(y_menu_test, menu_tree_predictions, average="weighted"),
        boost_f1_weighted
    ]
})

menu_model_comparison
````
