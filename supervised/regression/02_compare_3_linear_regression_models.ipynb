# Three Very Basic Linear Regression Models

This notebook builds and compares three Linear Regression models.

Only numerical columns are used. There is no imputation, encoding, pipeline, VIF, heatmap, or cross-validation.


## 1. Import libraries


```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

## 2. Load the dataset


```python
df = pd.read_csv("ecommerce_supervised_dataset.csv")
df.head()
```

## 3. Select only numerical columns


```python
numerical_data = df.select_dtypes(include=["int64", "float64"])
numerical_data.head()
```

## 4. Select the target


```python
y = numerical_data["Purchase_Amount"]
```

# Model 1: Original numerical features


```python
X1 = numerical_data.drop(columns=["Purchase_Amount"])
X1.head()
```

## Split, train, and predict


```python
X1_train, X1_test, y1_train, y1_test = train_test_split(
    X1, y, test_size=0.20, random_state=42
)

model_1 = LinearRegression()
model_1.fit(X1_train, y1_train)

predictions_1 = model_1.predict(X1_test)
```

## Model 1 accuracy check


```python
mae_1 = mean_absolute_error(y1_test, predictions_1)
mse_1 = mean_squared_error(y1_test, predictions_1)
rmse_1 = np.sqrt(mse_1)
r2_1 = r2_score(y1_test, predictions_1)

print("MAE :", mae_1)
print("MSE :", mse_1)
print("RMSE:", rmse_1)
print("R²  :", r2_1)
```

# Model 2: Add two squared features


```python
X2 = X1.copy()

X2["Previous_Purchases_Squared"] = X2["Previous_Purchases"] ** 2
X2["Session_Minutes_Squared"] = X2["Average_Session_Minutes"] ** 2

X2.head()
```

## Split, train, predict, and check accuracy


```python
X2_train, X2_test, y2_train, y2_test = train_test_split(
    X2, y, test_size=0.20, random_state=42
)

model_2 = LinearRegression()
model_2.fit(X2_train, y2_train)

predictions_2 = model_2.predict(X2_test)

mae_2 = mean_absolute_error(y2_test, predictions_2)
mse_2 = mean_squared_error(y2_test, predictions_2)
rmse_2 = np.sqrt(mse_2)
r2_2 = r2_score(y2_test, predictions_2)

print("MAE :", mae_2)
print("MSE :", mse_2)
print("RMSE:", rmse_2)
print("R²  :", r2_2)
```

# Model 3: Add simple interaction features


```python
X3 = X2.copy()

X3["Purchase_Engagement"] = (
    X3["Previous_Purchases"] * X3["Average_Session_Minutes"]
)

X3["Income_Per_Visit"] = (
    X3["Annual_Income"] /
    (X3["Website_Visits_Per_Month"] + 1)
)

X3["Age_Income_Interaction"] = (
    X3["Customer_Age"] * X3["Annual_Income"]
)

X3.head()
```

## Split, train, predict, and check accuracy


```python
X3_train, X3_test, y3_train, y3_test = train_test_split(
    X3, y, test_size=0.20, random_state=42
)

model_3 = LinearRegression()
model_3.fit(X3_train, y3_train)

predictions_3 = model_3.predict(X3_test)

mae_3 = mean_absolute_error(y3_test, predictions_3)
mse_3 = mean_squared_error(y3_test, predictions_3)
rmse_3 = np.sqrt(mse_3)
r2_3 = r2_score(y3_test, predictions_3)

print("MAE :", mae_3)
print("MSE :", mse_3)
print("RMSE:", rmse_3)
print("R²  :", r2_3)
```

# Compare the three models


```python
comparison = pd.DataFrame({
    "Model": ["Model 1", "Model 2", "Model 3"],
    "MAE": [mae_1, mae_2, mae_3],
    "MSE": [mse_1, mse_2, mse_3],
    "RMSE": [rmse_1, rmse_2, rmse_3],
    "R2": [r2_1, r2_2, r2_3]
})

comparison
```

## Best model


```python
best_model = comparison.sort_values("R2", ascending=False).iloc[0]

print("Best model:", best_model["Model"])
print("Best R² score:", best_model["R2"])
```
