# Seaborn Visualizations Using the Software Companies Dataset

This notebook contains **10 beginner-friendly Seaborn visualizations**.

Seaborn is built on Matplotlib and provides convenient functions for statistical charts.
Every example uses the supplied software companies CSV file.

## 1. Import Libraries and Load the Dataset

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

sns.set_theme(style="whitegrid")

csv_file = Path("software_companies_dataset_v2.csv")

if not csv_file.exists():
    raise FileNotFoundError(
        f"{csv_file} was not found. Place the CSV file in the same folder as the notebook."
    )

df = pd.read_csv(csv_file)

df["Incorporation_Date"] = pd.to_datetime(df["Incorporation_Date"], errors="coerce")
df["Last_Funding_Date"] = pd.to_datetime(df["Last_Funding_Date"], errors="coerce")

print("Dataset shape:", df.shape)
df.head()
```

## Dataset Overview

Review the available columns and missing values before plotting.

```python
print("Columns:")
print(df.columns.tolist())

print("\nMissing values:")
print(df.isna().sum())
```

## Visualization 1: Count Plot — Companies by Industry

A count plot displays how many rows belong to each category.  
This chart counts companies in every industry.

```python
plt.figure(figsize=(10, 5))
sns.countplot(
    data=df,
    x="Industry",
    order=df["Industry"].value_counts().index,
    color="steelblue"
)
plt.title("Number of Companies by Industry")
plt.xlabel("Industry")
plt.ylabel("Number of Companies")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

## Visualization 2: Bar Plot — Average Revenue by Ownership Type

A bar plot displays an estimated value for each category.  
Here, it shows the average annual revenue for each ownership type.

```python
plt.figure(figsize=(9, 5))
sns.barplot(
    data=df,
    x="Ownership_Type",
    y="Annual_Revenue",
    errorbar=None,
    color="steelblue"
)
plt.title("Average Annual Revenue by Ownership Type")
plt.xlabel("Ownership Type")
plt.ylabel("Average Annual Revenue")
plt.xticks(rotation=30, ha="right")
plt.tight_layout()
plt.show()
```

## Visualization 3: Histogram — Employee Satisfaction

A histogram shows the frequency distribution of numerical values.  
The optional curve provides a smooth view of the distribution.

```python
plt.figure(figsize=(9, 5))
sns.histplot(
    data=df,
    x="Employee_Satisfaction",
    bins=15,
    kde=True
)
plt.title("Distribution of Employee Satisfaction")
plt.xlabel("Employee Satisfaction")
plt.ylabel("Number of Companies")
plt.tight_layout()
plt.show()
```

## Visualization 4: Box Plot — Salary by Customer Segment

A box plot compares the distribution of a numerical variable across categories.  
This example compares average salary values across customer segments.

```python
plt.figure(figsize=(9, 5))
sns.boxplot(
    data=df,
    x="Customer_Segment",
    y="Average_Salary"
)
plt.title("Average Salary by Customer Segment")
plt.xlabel("Customer Segment")
plt.ylabel("Average Salary")
plt.tight_layout()
plt.show()
```

## Visualization 5: Violin Plot — Profit Margin by Risk Rating

A violin plot shows the shape and spread of a distribution.  
It compares profit margins for different risk ratings.

```python
plt.figure(figsize=(9, 5))
sns.violinplot(
    data=df,
    x="Risk_Rating",
    y="Profit_Margin",
    inner="box"
)
plt.title("Profit Margin by Risk Rating")
plt.xlabel("Risk Rating")
plt.ylabel("Profit Margin")
plt.tight_layout()
plt.show()
```

## Visualization 6: Scatter Plot — R&D Spending vs Annual Revenue

A scatter plot shows the relationship between two numerical variables.  
The point color represents the company's risk rating.

```python
plt.figure(figsize=(9, 6))
sns.scatterplot(
    data=df,
    x="R&D_Spending",
    y="Annual_Revenue",
    hue="Risk_Rating",
    alpha=0.7
)
plt.title("R&D Spending vs Annual Revenue")
plt.xlabel("R&D Spending")
plt.ylabel("Annual Revenue")
plt.tight_layout()
plt.show()
```

## Visualization 7: Line Plot — Average Cloud Adoption by Founding Decade

A line plot displays change across an ordered value.  
Companies are grouped into founding decades, and average cloud adoption is calculated.

```python
df["Founding_Decade"] = (df["Year_Founded"] // 10) * 10

decade_cloud = (
    df.groupby("Founding_Decade", as_index=False)["Adoption_Rate_Cloud"]
      .mean()
      .sort_values("Founding_Decade")
)

plt.figure(figsize=(9, 5))
sns.lineplot(
    data=decade_cloud,
    x="Founding_Decade",
    y="Adoption_Rate_Cloud",
    marker="o"
)
plt.title("Average Cloud Adoption by Founding Decade")
plt.xlabel("Founding Decade")
plt.ylabel("Average Cloud Adoption Rate")
plt.tight_layout()
plt.show()
```

## Visualization 8: Strip Plot — Product Lines by Ownership Type

A strip plot displays individual observations across categories.  
It helps us see how product-line counts vary within each ownership type.

```python
plt.figure(figsize=(10, 5))
sns.stripplot(
    data=df,
    x="Ownership_Type",
    y="Product_Lines",
    jitter=True,
    alpha=0.5
)
plt.title("Product Lines by Ownership Type")
plt.xlabel("Ownership Type")
plt.ylabel("Number of Product Lines")
plt.xticks(rotation=30, ha="right")
plt.tight_layout()
plt.show()
```

## Visualization 9: Heatmap — Correlation Between Numerical Columns

A heatmap represents correlation values using a colored grid.  
To keep it readable, this example uses a selected set of numerical columns.

```python
selected_columns = [
    "Employees",
    "Annual_Revenue",
    "Profit_Margin",
    "Market_Share",
    "R&D_Spending",
    "Average_Salary",
    "Employee_Satisfaction",
    "Adoption_Rate_AI",
    "Adoption_Rate_Cloud"
]

correlation = df[selected_columns].corr()

plt.figure(figsize=(10, 7))
sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f",
    cmap="coolwarm"
)
plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()
```

## Visualization 10: Pair Plot — Selected Business Metrics

A pair plot creates scatter plots for pairs of numerical columns and distributions on the diagonal.  
Only four columns are selected to keep the output easy to read.

```python
pair_columns = [
    "Annual_Revenue",
    "Profit_Margin",
    "Employee_Satisfaction",
    "Adoption_Rate_AI"
]

pair_data = df[pair_columns].dropna().sample(
    n=min(300, len(df)),
    random_state=42
)

sns.pairplot(pair_data, diag_kind="hist")
plt.suptitle("Pair Plot of Selected Business Metrics", y=1.02)
plt.show()
```
