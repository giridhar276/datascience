# Regular Expressions for Text Extraction and Masking

Python regular expressions are essential for deterministic text extraction, validation,
cleaning, and masking. They are often combined with NLP models in real-world pipelines.

> Run the installation/setup cell first. Some libraries may download language models or datasets the first time.

## Setup

```python
import re
import pandas as pd
```

## Use Case 1: Extract Email Addresses and Phone Numbers

```python
text = """
Contact Priya at priya@example.com or +91 98765 43210.
For escalation, write to support.team@company.org.
"""

emails = re.findall(r"\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}\b", text)
phones = re.findall(r"(?:\+91[\s-]?)?[6-9]\d{4}[\s-]?\d{5}", text)

print("Emails:", emails)
print("Phones:", phones)
```

## Use Case 2: Extract Case Numbers and Transaction IDs

```python
text = "Cases CS123456 and CS987654 refer to transactions TXN-2026-77881 and TXN-2026-88992."

cases = re.findall(r"\bCS\d{6}\b", text)
transactions = re.findall(r"\bTXN-\d{4}-\d{5}\b", text)

print(cases)
print(transactions)
```

## Use Case 3: Mask Sensitive Information

```python
def mask_sensitive(text):
    text = re.sub(
        r"\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}\b",
        "<EMAIL_ADDRESS>",
        text
    )
    text = re.sub(r"\bCS\d{6}\b", "<CASE_NUMBER>", text)
    text = re.sub(r"\b(?:\d{1,3}\.){3}\d{1,3}\b", "<IP_ADDRESS>", text)
    return text

sample = "Email me at user@example.com regarding CS123456 from IP 10.20.30.40."
print(mask_sensitive(sample))
```

## Use Case 4: Normalize Repeated Spaces and Punctuation

```python
noisy = "Payment   failed!!!   Please   help....."

cleaned = re.sub(r"\s+", " ", noisy)
cleaned = re.sub(r"([!?.,])\1+", r"\1", cleaned)

print(cleaned.strip())
```

## Use Case 5: Parse Semi-Structured Logs

```python
logs = [
    "2026-07-29 09:10:11 | ERROR | payment-service | timeout",
    "2026-07-29 09:11:05 | INFO | auth-service | login successful",
    "2026-07-29 09:12:40 | WARN | delivery-service | tracking delayed"
]

pattern = re.compile(
    r"(?P<timestamp>[\d-]+\s[\d:]+)\s\|\s"
    r"(?P<level>\w+)\s\|\s"
    r"(?P<service>[\w-]+)\s\|\s"
    r"(?P<message>.+)"
)

records = [match.groupdict() for line in logs if (match := pattern.match(line))]
pd.DataFrame(records)
```

## Use Case 6: Keyword Pattern Classification

```python
patterns = {
    "payment": re.compile(r"\b(payment|transaction|card|debited)\b", re.I),
    "login": re.compile(r"\b(login|password|sign[ -]?in|account locked)\b", re.I),
    "delivery": re.compile(r"\b(delivery|shipment|parcel|courier)\b", re.I),
    "refund": re.compile(r"\b(refund|reversal|money back)\b", re.I)
}

def classify(text):
    matches = [label for label, pattern in patterns.items() if pattern.search(text)]
    return matches or ["unclassified"]

for message in [
    "My card payment was declined",
    "The parcel is delayed",
    "I cannot sign in",
    "When will I get my money back?"
]:
    print(classify(message), "->", message)
```
