## Use Case 1: Tokenizing Support Tickets
Tokenization splits raw text into sentences and words. It is commonly used before classification,
search, summarization, and entity extraction.

```python
from nltk.tokenize import sent_tokenize, word_tokenize

ticket = (
    "The payment failed twice. I contacted support yesterday, "
    "but the issue is still unresolved. Please investigate urgently."
)

sentences = sent_tokenize(ticket)
words = word_tokenize(ticket)

print("Sentences:")
for sentence in sentences:
    print("-", sentence)

print("\nFirst 15 tokens:", words[:15])
```

## Use Case 2: Cleaning Text with Stop Words and Lemmatization
Stop-word removal reduces common words that may not help analysis. Lemmatization converts words
to their dictionary form, such as *running* to *run*.

```python
import re
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer

text = "Customers were complaining about delayed deliveries and damaged packages."
tokens = [w.lower() for w in word_tokenize(text) if re.fullmatch(r"[A-Za-z]+", w)]

stop_words = set(stopwords.words("english"))
lemmatizer = WordNetLemmatizer()

cleaned = [
    lemmatizer.lemmatize(word)
    for word in tokens
    if word not in stop_words
]

print("Original tokens:", tokens)
print("Cleaned tokens :", cleaned)
```

## Use Case 3: Part-of-Speech Tagging
POS tagging identifies nouns, verbs, adjectives, and other grammatical roles. It can help extract
product names, actions, issues, and descriptive words from feedback.

```python
from nltk import pos_tag

feedback = "The new mobile application loads slowly but provides useful payment alerts."
tokens = word_tokenize(feedback)
tagged = pos_tag(tokens)

print(tagged)
```

## Use Case 4: Named-Entity Recognition
Named-entity recognition identifies people, organizations, and locations. This is helpful for
routing documents, compliance checks, and extracting key details from messages.

```python
from nltk import ne_chunk

sentence = "Anita from Infosys visited Hyderabad on Monday to meet the analytics team."
tokens = word_tokenize(sentence)
tagged = pos_tag(tokens)
tree = ne_chunk(tagged)

print(tree)
```

## Use Case 5: N-gram Analysis
N-grams reveal commonly occurring phrases. Bigram and trigram analysis can identify repeated
customer concerns such as *payment failure* or *delivery delay*.

```python
from nltk.util import ngrams
from collections import Counter

reviews = [
    "payment failed during checkout",
    "payment failed again today",
    "delivery delay caused inconvenience",
    "delivery delay is unacceptable",
    "checkout payment failed"
]

all_tokens = []
for review in reviews:
    all_tokens.extend([
        word.lower() for word in word_tokenize(review)
        if word.isalpha()
    ])

bigrams = Counter(ngrams(all_tokens, 2))
print("Most common bigrams:")
for phrase, count in bigrams.most_common(8):
    print(" ".join(phrase), "->", count)
```

## Use Case 6: Sentiment Analysis with VADER
VADER is a rule-based sentiment analyzer that performs well on short messages and social-media-like
text. The compound score ranges from -1 to +1.

```python
from nltk.sentiment import SentimentIntensityAnalyzer

sia = SentimentIntensityAnalyzer()
messages = [
    "The service was excellent and the issue was resolved quickly.",
    "The product is okay, but delivery took longer than expected.",
    "This is terrible. My account is still blocked!"
]

for message in messages:
    score = sia.polarity_scores(message)
    print(message)
    print(score, "\n")
```
