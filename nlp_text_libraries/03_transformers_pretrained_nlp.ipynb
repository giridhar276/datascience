# Hugging Face Transformers for Pretrained NLP

Transformers provides access to pretrained language models for sentiment analysis,
zero-shot classification, named-entity recognition, summarization, question answering, and text generation.
The examples download models from Hugging Face when first executed.



## Setup

```python
# Uncomment if needed
# %pip install transformers torch sentencepiece

from transformers import pipeline
```

## Use Case 1: Sentiment Analysis
A pretrained model can classify sentiment without training a model from scratch.

```python
sentiment = pipeline("sentiment-analysis")

texts = [
    "The service was excellent and the issue was resolved.",
    "The application crashes every time I log in."
]

sentiment(texts)
```

## Use Case 2: Zero-Shot Ticket Classification
Zero-shot classification assigns labels that were not part of model-specific training code.

```python
classifier = pipeline(
    "zero-shot-classification",
    model="facebook/bart-large-mnli"
)

ticket = "My parcel has not arrived and tracking is not updating."
labels = ["payment issue", "login issue", "delivery issue", "refund issue"]

classifier(ticket, candidate_labels=labels)
```

## Use Case 3: Named-Entity Recognition

```python
ner = pipeline(
    "token-classification",
    model="dslim/bert-base-NER",
    aggregation_strategy="simple"
)

text = "Ravi from IBM travelled to Singapore for a client workshop."
ner(text)
```

## Use Case 4: Summarization

```python
summarizer = pipeline(
    "summarization",
    model="sshleifer/distilbart-cnn-12-6"
)

long_text = (
    "The customer reported that the transaction failed during checkout. "
    "Although the application displayed an error, the amount was debited. "
    "The customer contacted support and shared the transaction reference. "
    "The support team confirmed that the payment gateway timed out and initiated a reversal. "
    "The customer was informed that the amount should be credited within five business days."
)

summarizer(long_text, max_length=60, min_length=20, do_sample=False)
```

## Use Case 5: Extractive Question Answering

```python
qa = pipeline(
    "question-answering",
    model="distilbert-base-cased-distilled-squad"
)

context = (
    "The refund was initiated on Monday. The bank normally takes "
    "five business days to credit the amount."
)

qa(
    question="How long will the refund take?",
    context=context
)
```

## Use Case 6: Batch Processing
Batching improves throughput when analyzing multiple texts.

```python
batch = [
    "The user interface is simple.",
    "The shipment was delayed.",
    "Support solved my problem quickly.",
    "The payment page is confusing."
]

sentiment(batch, batch_size=4)
```
